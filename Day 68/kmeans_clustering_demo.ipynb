```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv('/content/student_clustering.csv')
```

```python
df.shape
```

```python
df.head()
```

```python
import matplotlib.pyplot as plt

plt.scatter(x=df['cgpa'],y=df['iq'])
```

```python
from sklearn.cluster import KMeans
```

```python
wcss=[]

for i in range(1,11):
  km=KMeans(n_clusters=i)
  km.fit_predict(df)
  wcss.append(km.inertia_)
```

```python
wcss
```

```python
plt.plot(range(1,11),wcss)
```

```python
X=df.iloc[:,:].values
km=KMeans(n_clusters=4)
y_mean = km.fit_predict(X)
```

```python
y_mean
```

```python
X[y_mean==0,1]
```

```python
plt.scatter(X[y_mean==0,0],X[y_mean==0,1],color='r')
plt.scatter(X[y_mean==1,0],X[y_mean==1,1],color='b')
plt.scatter(X[y_mean==2,0],X[y_mean==2,1],color='g')
plt.scatter(X[y_mean==3,0],X[y_mean==3,1],color='black')
```

```python
from sklearn.datasets import make_blobs

centroids = [(-5,-5,5),(5,5,-5),(3.5,-2.5,4),(-2.5,2.5,-4)]
cluster_std = [1,1,1,1]

X,y = make_blobs(n_samples=200,cluster_std=cluster_std,centers=centroids,n_features=3,random_state=1)
```

```python
X
```

```python
import plotly.express as px
fig=px.scatter_3d(x=X[:,0],y=X[:,1],z=X[:,2])
fig.show()
```
